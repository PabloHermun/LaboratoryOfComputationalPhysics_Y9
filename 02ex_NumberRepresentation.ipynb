1\. Write a function that converts number representation, bin<->dec<->hex. (Clearly using the corresponding python built-in functions is not fair..)

```python
def my_bin(n: int):
    bin_rep = [0] # binary representation (reversed)
    
    # Case n=0
    if n == 0:
        return str(bin_rep)
        
    # Case n>0 
    # smaller digit
    if n%2 == 1:
        bin_rep[0] += 1

    # middle digits
    i = 1
    while 2**i < abs(n): #           
        print(2**i, n%(2**i)) 
        bin_rep.append(n%2**i) 
        i+=1

    # bigger digit (if n is power of 2)
    if abs(n) == 2**i:
        bin_rep.append(1)

    # Append sign
    if n < 0:
        bin_rep.append('-')
        
    print(bin_rep) #DEBUGGING
    bin_rep.reverse()
    return bin_rep
```
